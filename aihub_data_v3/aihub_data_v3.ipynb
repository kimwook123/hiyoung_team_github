```python
import os
import shutil

# 1. 경로 설정
dataset_1_path = '/content/dataset_1/aihub_dataset' # 5만장 폴더
dataset_2_path = '/content/dataset_2/dataset'      # 4천장 폴더
output_path = '/content/dataset'             # 합쳐질 폴더

# 폴더 구조 생성
for folder in ['images/train', 'images/val', 'labels/train', 'labels/val']:
    os.makedirs(os.path.join(output_path, folder), exist_ok=True)

def copy_files(src_root, dst_root):
    # images와 labels 각각 복사
    for mode in ['train', 'val']:
        src_img = os.path.join(src_root, 'images', mode)
        src_lbl = os.path.join(src_root, 'labels', mode)
        dst_img = os.path.join(dst_root, 'images', mode)
        dst_lbl = os.path.join(dst_root, 'labels', mode)

        # 이미지 복사
        if os.path.exists(src_img):
            for file in os.listdir(src_img):
                shutil.copy(os.path.join(src_img, file), os.path.join(dst_img, file))

        # 라벨 복사
        if os.path.exists(src_lbl):
            for file in os.listdir(src_lbl):
                shutil.copy(os.path.join(src_lbl, file), os.path.join(dst_lbl, file))

print("데이터셋 합치는 중...")
copy_files(dataset_1_path, output_path)
copy_files(dataset_2_path, output_path)
print(f"완료! 최종 데이터셋은 {output_path}에 생성되었습니다.")
```

```python
import yaml

# 데이터셋 구조에 맞는 경로 설정
data_config = {
    'path': '/content/dataset',
    'train': 'images/train',
    'val': 'images/val',
    'names': {
        0: 'YES_Helmet',
        1: 'NO_Helmet'
    }
}

# YAML 파일로 저장
with open('/content/dataset/data.yaml', 'w') as f:
    yaml.dump(data_config, f, default_flow_style=False)

print("data.yaml 파일이 생성되었습니다!")
```

```python
import os
import glob

# 하위 폴더까지 모두 검색하기 위해 recursive=True와 ** 사용
train_labels = glob.glob('/content/dataset/labels/train/**/*.txt', recursive=True)
valid_labels = glob.glob('/content/dataset/labels/valid/**/*.txt', recursive=True)
all_labels = train_labels + valid_labels

class_counts = {0: 0, 1: 0}
total_boxes = 0

print(f"총 {len(all_labels)}개의 라벨 파일 분석 시작...")

for label_path in all_labels:
    try:
        with open(label_path, 'r') as f:
            for line in f:
                parts = line.strip().split()
                if not parts: continue

                class_id = int(parts[0])
                if class_id in class_counts:
                    class_counts[class_id] += 1
                    total_boxes += 1
    except Exception as e:
        print(f"Error reading {label_path}: {e}")

print("=" * 50)
print(f"[라벨 전수조사 결과] 총 라벨 파일 수: {len(all_labels)}개")
print(f"전체 Bounding Box 개수: {total_boxes}개")
print("-" * 50)
print(f"Class 0 (YES_Helmet) 개수 : {class_counts[0]:,}개")
print(f"Class 1 (NO_Helmet) 개수   : {class_counts[1]:,}개")
print("=" * 50)

# 클래스 분포 비율 확인
if total_boxes > 0:
    ratio0 = (class_counts[0] / total_boxes) * 100
    ratio1 = (class_counts[1] / total_boxes) * 100
    print(f"분포 비율 -> YES: {ratio0:.1f}%, NO: {ratio1:.1f}%")
```

```python
from ultralytics import YOLO

model = YOLO('best.pt')

# 파인튜닝 및 체크포인트 설정
model.train(
    data='/content/dataset/data.yaml',
    optimizer='MuSGD',
    epochs=50,
    imgsz=640,
    batch=128,
    workers=8,
    lr0=0.02,
    lrf=0.01,
    momentum=0.937,
    weight_decay=0.0005,
    freeze=0,
    cls=2.5,                   # 편향 해결을 위한 가중치
    save_period=5,             # 5 에폭마다 모델을 자동으로 저장합니다.
    mosaic=1.0,
    mixup=0.1,
)
```
