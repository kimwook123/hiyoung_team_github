```python
def fix_labels_to_person(label_folder):
    if not os.path.exists(label_folder):
        return

    for f in os.listdir(label_folder):
        if f.endswith('.txt'):
            path = os.path.join(label_folder, f)
            with open(path, 'r') as file:
                lines = file.readlines()

            with open(path, 'w') as file:
                for line in lines:
                    parts = line.split()
                    if parts:
                        # 무조건 1번(person)으로 변경
                        parts[0] = '1'
                        file.write(" ".join(parts) + "\n")

# train과 valid 폴더 모두 적용
fix_labels_to_person('/content/new_data/train/labels')
fix_labels_to_person('/content/new_data/valid/labels')

print("✅ 새 데이터의 모든 라벨을 '1번(person)'으로 변경 완료!")
```

```python
import shutil

def final_merge(src_root, dst_root, prefix='new_'):
    for split in ['train', 'valid']:
        src_img_dir = os.path.join(src_root, split, 'images')
        src_lbl_dir = os.path.join(src_root, split, 'labels')
        dst_img_dir = os.path.join(dst_root, split, 'images')
        dst_lbl_dir = os.path.join(dst_root, split, 'labels')

        if not os.path.exists(src_img_dir): continue

        images = [f for f in os.listdir(src_img_dir) if f.lower().endswith(('.jpg', '.png', '.jpeg'))]
        for img in images:
            # 파일 복사 (이름 변경 포함)
            shutil.copy(os.path.join(src_img_dir, img), os.path.join(dst_img_dir, prefix + img))

            # 라벨 복사
            lbl = os.path.splitext(img)[0] + '.txt'
            if os.path.exists(os.path.join(src_lbl_dir, lbl)):
                shutil.copy(os.path.join(src_lbl_dir, lbl), os.path.join(dst_lbl_dir, prefix + lbl))

    print(f"🚀 모든 데이터가 '{dst_root}' 폴더로 합쳐졌습니다!")

# 실행
final_merge('/content/new_data', '/content/dataset_v4')
```

```python
import os
import random
import shutil

# 경로 설정
base_path = '/content/dataset_v4'
train_img_dir = os.path.join(base_path, 'train/images')
train_lbl_dir = os.path.join(base_path, 'train/labels')

# 새 폴더 생성 (valid, test)
for split in ['valid', 'test']:
    os.makedirs(os.path.join(base_path, split, 'images'), exist_ok=True)
    os.makedirs(os.path.join(base_path, split, 'labels'), exist_ok=True)

# 전체 이미지 목록 가져오기
images = [f for f in os.listdir(train_img_dir) if f.lower().endswith(('.jpg', '.jpeg', '.png'))]
random.shuffle(images) # 무작위로 섞기

# 개수 계산 (10%씩)
num_total = len(images)
num_val = int(num_total * 0.1)
num_test = int(num_total * 0.1)

val_files = images[:num_val]
test_files = images[num_val : num_val + num_test]

def move_files(file_list, target_split):
    for img_name in file_list:
        # 이미지 이동
        shutil.move(os.path.join(train_img_dir, img_name),
                    os.path.join(base_path, target_split, 'images', img_name))
        # 라벨 이동
        lbl_name = os.path.splitext(img_name)[0] + '.txt'
        src_lbl = os.path.join(train_lbl_dir, lbl_name)
        if os.path.exists(src_lbl):
            shutil.move(src_lbl, os.path.join(base_path, target_split, 'labels', lbl_name))

# 실행
move_files(val_files, 'valid')
move_files(test_files, 'test')

print(f"✅ 분할 완료!")
print(f"📊 Train: {len(os.listdir(train_img_dir))}장")
print(f"📊 Valid: {len(val_files)}장")
print(f"📊 Test: {len(test_files)}장")
```

```python
import os

base_path = '/content/dataset_v4'
splits = ['train', 'valid', 'test']

print("=" * 50)
print("🚀 [데이터셋 검증 결과] 🚀")
print("=" * 50)

for split in splits:
    img_dir = os.path.join(base_path, split, 'images')
    lbl_dir = os.path.join(base_path, split, 'labels')

    # 폴더 존재 여부 확인
    if not os.path.exists(img_dir) or not os.path.exists(lbl_dir):
        print(f"⚠️ [{split.upper()}] 폴더가 존재하지 않거나 경로가 잘못되었습니다.")
        continue

    # 이미지 및 라벨 개수 확인
    images = [f for f in os.listdir(img_dir) if f.lower().endswith(('.jpg', '.jpeg', '.png'))]
    labels = [f for f in os.listdir(lbl_dir) if f.endswith('.txt')]

    # 클래스별 카운트 초기화 (0: 헬멧, 1: 사람, 2: 조끼)
    class_counts = {'0': 0, '1': 0, '2': 0}

    # 라벨 파일 읽어서 클래스 개수 세기
    for lbl in labels:
        with open(os.path.join(lbl_dir, lbl), 'r') as f:
            for line in f:
                parts = line.split()
                if parts:
                    class_id = parts[0]
                    if class_id in class_counts:
                        class_counts[class_id] += 1
                    else:
                        # 지정된 0,1,2 외에 다른 클래스 번호가 섞여있을 경우 처리
                        class_counts[class_id] = class_counts.get(class_id, 0) + 1

    print(f"📂 분할 섹션: [{split.upper()}]")
    print(f"🖼️  이미지 수: {len(images)}장 | 📝 라벨 파일 수: {len(labels)}장")
    print(f"🎯 클래스별 박스 현황:")
    print(f"   - [0] 헬멧(Helmet) : {class_counts.get('0', 0)}개")
    print(f"   - [1] 사람(Person) : {class_counts.get('1', 0)}개")
    print(f"   - [2] 조끼(Vest)   : {class_counts.get('2', 0)}개")

    # 만약 정체불명의 클래스가 검출되었다면 경고 표시
    extra_classes = {k: v for k, v in class_counts.items() if k not in ['0', '1', '2']}
    if extra_classes:
        print(f"   🚨 경고! 정의되지 않은 클래스 번호 검출됨: {extra_classes}")

    print("-" * 50)
```

```python
import os
import random
import shutil

base_path = '/content/dataset_v4'
train_img_dir = os.path.join(base_path, 'train/images')
train_lbl_dir = os.path.join(base_path, 'train/labels')

# 1. 새 폴더 생성 (valid, test)
for split in ['valid', 'test']:
    os.makedirs(os.path.join(base_path, split, 'images'), exist_ok=True)
    os.makedirs(os.path.join(base_path, split, 'labels'), exist_ok=True)

# 2. 전체 이미지 목록 정렬 후 무작위 섞기 (어제와 동일한 분포를 유도하기 위해 seed 고정)
images = [f for f in os.listdir(train_img_dir) if f.lower().endswith(('.jpg', '.jpeg', '.png'))]
images.sort() # 순서 정렬
random.seed(42) # 어제 설정했던 시드 42 고정!
random.shuffle(images)

# 3. 개수 계산 (10%씩)
num_total = len(images)
num_val = int(num_total * 0.1)
num_test = int(num_total * 0.1)

val_files = images[:num_val]
test_files = images[num_val : num_val + num_test]

def move_files(file_list, target_split):
    for img_name in file_list:
        shutil.move(os.path.join(train_img_dir, img_name),
                    os.path.join(base_path, target_split, 'images', img_name))
        lbl_name = os.path.splitext(img_name)[0] + '.txt'
        src_lbl = os.path.join(train_lbl_dir, lbl_name)
        if os.path.exists(src_lbl):
            shutil.move(src_lbl, os.path.join(base_path, target_split, 'labels', lbl_name))

# 4. 실행
move_files(val_files, 'valid')
move_files(test_files, 'test')

print("✅ 어제와 동일한 기준으로 Valid / Test 분할 완료!")
print(f"📊 Train: {len(os.listdir(train_img_dir))}장 | Valid: {len(val_files)}장 | Test: {len(test_files)}장")
```

```python
import os

label_dirs = [
    '/content/drive/MyDrive/project_v5/dataset/labels/train',
    '/content/drive/MyDrive/project_v5/dataset/labels/val'
]

print("🛡️ 오염 방지 안전 필터 적용 - person(Class 2) 주입 시작...")

for label_dir in label_dirs:
    if not os.path.exists(label_dir):
        print(f"❌ 경로를 찾을 수 없습니다: {label_dir}")
        continue

    for filename in os.listdir(label_dir):
        if not filename.endswith('.txt'):
            continue

        file_path = os.path.join(label_dir, filename)

        with open(file_path, 'r', encoding='utf-8') as f:
            lines = f.readlines()

        clean_lines = []

        for line in lines:
            parts = line.strip().split()
            if not parts or len(parts) < 5:
                continue # 정상적인 5개 요소(클래스,x,y,w,h)가 아니면 과감히 버림

            cls = int(parts[0])
            # 혹시나 기존에 꼬여서 들어가 있던 2번(person)이나 깨진 문자열은 과감히 무시하고
            # 오직 원본의 순수한 0번(helmet)과 1번(head)만 깨끗하게 수집합니다.
            if cls in [0, 1]:
                x, y, w, h = float(parts[1]), float(parts[2]), float(parts[3]), float(parts[4])

                # 원본 라벨 복구 및 강제 줄바꿈 세팅
                clean_lines.append(f"{cls} {x:.6f} {y:.6f} {w:.6f} {h:.6f}\n")

                # [YOLO 상단 원점 좌표계 반영한 전신 계산]
                person_w = min(w * 3.0, 0.95)
                person_h = min(h * 6.5, 0.95)

                head_top = y - (h / 2.0) # 정수리 위치 구하기
                person_y = head_top + (person_h / 2.0) # 아래로 전신 박스 내리기

                # 이미지 경계선 방어선
                if person_y - (person_h / 2.0) < 0: person_y = person_h / 2.0
                if person_y + (person_h / 2.0) > 1.0: person_y = 1.0 - (person_h / 2.0)

                clean_lines.append(f"2 {x:.6f} {person_y:.6f} {person_w:.6f} {person_h:.6f}\n")

        # 완전히 새로 정제된 내용으로 덮어쓰기 (기존 오염물 전체 삭제 효과)
        with open(file_path, 'w', encoding='utf-8') as f:
            f.writelines(clean_lines)

print("🎉 원본 헬멧/헤드 완벽 복구 및 정상적인 전신 박스 합성 완료!")
```

```python
import cv2
import yt_dlp
from ultralytics import YOLO
import os

# 1. 테스트하고 싶은 유튜브 URL 및 영상 저장 경로
youtube_url = "https://youtu.be/Uyn9k9kKkT4?si=-ahk3_4pygpczgj2"
video_save_path = "/content/downloaded_video.mp4"

# 2. yt-dlp로 유튜브 영상 다운로드
ydl_opts = {
    'format': 'bestvideo[ext=mp4]+bestaudio[ext=m4a]/best[ext=mp4]/best',
    'outtmpl': video_save_path,
    'quiet': True
}

print("📥 유튜브 영상 다운로드 중...")
with yt_dlp.YoutubeDL(ydl_opts) as ydl:
    ydl.download([youtube_url])
print("✅ 다운로드 완료!")

# 3. 코랩 파일 서랍에 올린 두 개의 가중치 모델 로드
# 경로가 /content/person_detect.pt 인지 다시 한번 확인해 주세요!
person_model_path = "/content/person_detect.pt"
safety_model_path = "/content/good1.pt"

if not os.path.exists(person_model_path) or not os.path.exists(safety_model_path):
    raise FileNotFoundError("❌ 파일 서랍에 person_detect.pt와 good1.pt가 올바르게 업로드되었는지 확인해 주세요!")

person_model = YOLO(person_model_path)
safety_model = YOLO(safety_model_path)
print("🎯 두 개의 모델(사람 탐지 + 안전모 탐지) 로드 완료!")

# 4. 영상 처리를 위한 OpenCV 설정
cap = cv2.VideoCapture(video_save_path)
width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
fps = int(cap.get(cv2.CAP_PROP_FPS)) if cap.get(cv2.CAP_PROP_FPS) else 30
total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))

output_path = '/content/output_ensemble_test.mp4'
fourcc = cv2.VideoWriter_fourcc(*'mp4v')
out = cv2.VideoWriter(output_path, fourcc, fps, (width, height))

print(f"🚀 [전체 {total_frames}개 프레임] 앙상블 기반 모든 프레임 변환 시작...")
frame_count = 0

while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        break

    # 1. 두 모델 각각 독립적으로 추론 실행
    person_results = person_model(frame, conf=0.40, verbose=False)
    safety_results = safety_model(frame, conf=0.40, verbose=False)

    # 2. 화면을 그리기 전, 먼저 안전모 모델의 결과로 프레임을 1차 플로팅(렌더링)합니다.
    # 이렇게 하면 헬멧과 맨머리 박스가 먼저 프레임 위에 예쁘게 그려집니다.
    annotated_frame = safety_results[0].plot()

    # 3. 사람 탐지 모델의 결과 창고를 열어봅니다.
    boxes = person_results[0].boxes
    if len(boxes) > 0:
        # 안전한 수정을 위해 GPU/CPU 메모리에 있는 텐서 데이터를 완전히 복제(Clone)합니다.
        # 이렇게 하면 Read-Only 잠금이 풀리고 자유롭게 수정할 수 있게 됩니다.
        cloned_boxes = boxes.data.clone()

        # 4. 사람 모델이 찾은 박스들의 클래스 번호를 강제로 2번(person)으로 변경합니다.
        # YOLO의 박스 데이터 구조상 맨 마지막(5번째) 인덱스가 클래스 번호(cls)입니다.
        cloned_boxes[:, 5] = 2

        # 5. 클래스 번호가 2번으로 리맵핑된 사람 박스들만 골라서
        # 아까 안전모가 그려진 프레임 위에 2차로 덮어씌워 그려줍니다.
        for box in cloned_boxes:
            x1, y1, x2, y2, conf, cls = box

            # 박스 좌표를 정수형 변환
            p1, p2 = (int(x1), int(y1)), (int(x2), int(y2))

            # 사람 전신 박스 그리기 (파란색 박스, 두께 3)
            cv2.rectangle(annotated_frame, p1, p2, (255, 0, 0), 3)

            # 박스 상단에 'person'과 확률 텍스트 얹기
            label = f"person {conf:.2f}"
            cv2.putText(annotated_frame, label, (p1[0], p1[1] - 10),
                        cv2.FONT_HERSHEY_SIMPLEX, 0.6, (255, 0, 0), 2)

    # 최종 합성된 프레임을 영상 파일에 차곡차곡 저장
    out.write(annotated_frame)

    frame_count += 1
    if frame_count % 50 == 0:
        progress = (frame_count / total_frames) * 100
        print(f"🎬 앙상블 영상 굽는 중... [{frame_count}/{total_frames}] ({progress:.1f}%) 완료")

# 메모리 해제
cap.release()
out.release()
print(f"\n🎉 앙상블 추론 및 100% 영상 변환 완료!")
print(f"📂 최종 결과 동영상 다운로드 경로 ➡️ {output_path}")
```
